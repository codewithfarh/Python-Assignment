B 1. Grade calculator with nested conditions

Qno : 1 . write a program that takes astudent's makrs (0 - 100) and assigns a grade and remarks :
. 90 -100 : A+ outstanding | 80- 89 : A Excellent  | 70- 79 B Good 
. 60 -69 : C Average | 50-59 D Pass | Below 50 : F Fail 

additionally :
1. if makrs are below 0 or above 100, print an error and exit
2. print the grade and remarks
3. if failed, print : "Minimu marks to pass :50.you need x more marks

```python

marks = float(input("Enter student marks (0-100): "))


if marks < 0 or marks > 100:
    print("Error: Invalid marks! Please enter a value between 0 and 100.")
else:
    if marks >= 90:
        grade = "A+"
        remarks = "Outstanding"
    elif marks >= 80:
        grade = "A"
        remarks = "Excellent"
    elif marks >= 70:
        grade = "B"
        remarks = "Good"
    elif marks >= 60:
        grade = "C"
        remarks = "Average"
    elif marks >= 50:
        grade = "D"
        remarks = "Pass"
    else:
        grade = "F"
        remarks = "Fail"

  
    print(f"Grade: {grade}")
    print(f"Remarks: {remarks}")


    if grade == "F":
        needed_marks = 50 - marks
        print(f"Minimum marks to pass: 50. You need {needed_marks} more marks.")
```

pattern printing with nested loops

Q no : 2. using nested loop ,printthe following patterns. Accept n from input(use n = 5 for demo)
part 1 right triangle

part 2 multiplication table grid

```python
# Accept n from user
n = int(input("Enter the value of n (use 5 for demo): "))

print("\n--- Part 1: Right Triangle ---")
# Outer loop for number of rows
for i in range(1, n + 1):
    # Inner loop to print '*' in each row
    for j in range(i):
        print("*", end=" ")
    # Move to the next line after each row
    print()


print("\n--- Part 2: Multiplication Table Grid ---")
# Outer loop for rows (1 to n)
for row in range(1, n + 1):
    # Inner loop for columns (1 to n)
    for col in range(1, n + 1):
        # Calculate product
        product = row * col
        # Use end="\t" to create a clean tab-spaced grid
        print(product, end="\t")
    # Move to the next line after finishing a row
    print()
```

Q no : 3. Number analysis loop 
. Write a program that ask the user to enter numbers one by one. The user types "done" to Stop. After the loop, display :
1. Total count of numbers entered
2. Sum and average (rounded to 2 decimal places)
3. largest and smallest numbers 
4. Count of even vs odd numbers

handle the case where user enters a non-numeric value - Skip it and print a warning, but don't crash

```python

numbers = []
even_count = 0
odd_count = 0

print("Enter numbers one by one. Type 'done' to finish.")

while True:
    user_input = input("Enter a number (or 'done'): ").lower()
    
    
    if user_input == "done":
        break
  
    try:
       
        num = float(user_input)
        numbers.append(num)
        
      
        if num % 2 == 0:
            even_count += 1
        else:
            odd_count += 1
            
    except ValueError:
        print(f"Warning: '{user_input}' is not a valid number. Skipping it.")


if len(numbers) > 0:
    total_sum = sum(numbers)
    total_count = len(numbers)
    average = total_sum / total_count
    
    print("\n--- Number Analysis Results ---")
    print(f"1. Total count of numbers: {total_count}")
    print(f"2. Sum: {total_sum} | Average: {round(average, 2)}")
    print(f"3. Largest: {max(numbers)} | Smallest: {min(numbers)}")
    print(f"4. Even numbers: {even_count} | Odd numbers: {odd_count}")
else:
    print("No valid numbers were entered.")
```

Q no ; 4. print numbers from 1 to 100 applyonh these rules:
. Divisible by 3 --> Fizz
.  Divisible by 5 --> Buzz
.  Divisible by both 3 and 5 --> FizzBuzz
.  Divisible by 7 --> append Boom (or print Boom alone if no other label)
. Other wise print numbers

after printing all 100 numbers , print a summary showing how many times each label appeared.

```python
# 1. Initialize counters for the final summary
fizz_count = 0
buzz_count = 0
fizzbuzz_count = 0
boom_count = 0

# 2. Use a 'for' loop to iterate from 1 to 100
# This loop is safe and will stop automatically
for i in range(1, 101):
    output = ""
    
    # Check for 3 and 5 first (Specific rule)
    if i % 3 == 0 and i % 5 == 0:
        output = "FizzBuzz"
        fizzbuzz_count += 1
    elif i % 3 == 0:
        output = "Fizz"
        fizz_count += 1
    elif i % 5 == 0:
        output = "Buzz"
        buzz_count += 1
        
    # Check for 7 separately (To append "Boom")
    if i % 7 == 0:
        if output == "":
            output = "Boom"
        else:
            output = output + "Boom"
        boom_count += 1

    # 3. Print the label or the number if no rules matched
    if output == "":
        print(i)
    else:
        print(output)

# 4. Print the summary report (Outside the loop)
print("\n--- FINAL SUMMARY ---")
print(f"Fizz: {fizz_count}")
print(f"Buzz: {buzz_count}")
print(f"FizzBuzz: {fizzbuzz_count}")
print(f"Boom: {boom_count}")
```
