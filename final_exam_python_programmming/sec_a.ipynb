Q no: 2. write a program that accept a sentence from the user and performs these operations, printing each result :
1. count the total number of  words
2. count how many times the letter 'a' appears (case-sensitive)
3. Reverse the entire string
4. check if it is a palindrome  (ignoring spaces and case)
5. replace every vowel( a e i o u ) with *
 Test your program with : "Never odd or even"

```python
user = input("enter a sentence ")

print(user.count("a"))

word_in_list = user.split()

print(f"The number of word in sentence : {word_in_list}")

reverse_string = user[::-1]

print(f"The orginal sentence : {user}")
print(f"The reverse string :{reverse_string}")

if user .lower() == reverse_string.lower():
    print("Result: it is a palindrome")
else:
    print("Result: no it is not palindrome")

new_sentence = user.replace("a", "*").replace("e", "*").replace("i", "*").replace("O", "*").replace("u", "*")

print(f"after modifications the new sentence is : {new_sentence}")
```

Q no : 3. A shopkeeper needs a billing calculator: write a program that he takes :
. item price (float) Quantity(int),Discount % (float), tax rate (float)
calculate and print:
1. subtotal (price x quantity)
2. Discount amount
3. Amount after discount
4. tax amount (pakistan GST 17 %)
5. Grand total (rounded 2 decimal places)


```python
price = float(input("Enter item price: "))
quantity = int(input("Enter quantity: "))
discount_percent = float(input("Enter discount percentage (e.g. 10 for 10%): "))
tax_rate = 17.0  

subtotal = price * quantity


discount_amount = (subtotal * discount_percent) / 100


amount_after_discount = subtotal - discount_amount


tax_amount = (amount_after_discount * tax_rate) / 100


grand_total = amount_after_discount + tax_amount


print("-" * 30)
print(f"1. Subtotal: {subtotal}")
print(f"2. Discount Amount: {discount_amount}")
print(f"3. Amount After Discount: {amount_after_discount}")
print(f"4. Tax Amount (17% GST): {tax_amount}")
print(f"5. Grand Total: {round(grand_total, 2)}") 
print("-" * 30)
```

Q no : 4. write a program that collects and validates a Pakitani phone number. Rules :
. Must be exactly 11 digits 
. Must start with 03
. Must contain only digits(no spaces,dashes, or letters)

The program should be :
1. Ask the user to enter a phone number
2. Check all three rules using string method only (no import re)
3. print a clear message for each failed rules
4. if valid, print: Valid phone number :0234-3457652 (with dashes after position 4)

```python

phone_number = input("Enter your phone number: ")

is_valid = True


if len(phone_number) != 11:
    print("Failed: The number must be exactly 11 digits.")
    is_valid = False


if not phone_number.startswith("03"):
    print("Failed: The number must start with '03'.")
    is_valid = False


if not phone_number.isdigit():
    print("Failed: The number must contain only digits.")
    is_valid = False


if is_valid:
   
    formatted_number = phone_number[:4] + "-" + phone_number[4:]
    print("Valid phone number:", formatted_number)
else:
    print("Status: Invalid number. Please fix the errors above.")
```
