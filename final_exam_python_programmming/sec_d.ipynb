Q no : 1. You have a dictionary of products in a store:
inventory = {
    "Laptop":     {"price": 85000, "qty": 10, "category": "Electronics"},
    "T-Shirt":    {"price": 1200,  "qty": 50, "category": "Clothing"},
    "Headphones": {"price": 5500,  "qty": 20, "category": "Electronics"},
    "Jeans":      {"price": 3200,  "qty": 30, "category": "Clothing"},
    "Mouse":      {"price": 1800,  "qty": 15, "category": "Electronics"},
}

Write code to:
•	Print the total inventory value (price x qty for each, summed)
•	Find and print the most expensive item
•	List all items in the "Electronics" category
•	Apply a 10% discount to all items where price > 5000 (update dictionary)
•	Print a sorted list of items by price (ascending)

```python
inventory = {
    "Laptop":     {"price": 85000, "qty": 10, "category": "Electronics"},
    "T-Shirt":    {"price": 1200,  "qty": 50, "category": "Clothing"},
    "Headphones": {"price": 5500,  "qty": 20, "category": "Electronics"},
    "Jeans":      {"price": 3200,  "qty": 30, "category": "Clothing"},
    "Mouse":      {"price": 1800,  "qty": 15, "category": "Electronics"},
}


total_value = sum(item["price"] * item["qty"] for item in inventory.values())
print(f"Total Inventory Value: {total_value}")


most_expensive = max(inventory, key=lambda x: inventory[x]["price"])
print(f"Most Expensive Item: {most_expensive}")


electronics_items = [name for name, details in inventory.items() if details["category"] == "Electronics"]
print(f"Electronics Items: {electronics_items}")


for name, details in inventory.items():
    if details["price"] > 5000:
        details["price"] = details["price"] * 0.9 
print("Discount applied to eligible items.")


sorted_inventory = sorted(inventory.items(), key=lambda x: x[1]["price"])

print("\n--- Items Sorted by Price ---")
for name, details in sorted_inventory:
    print(f"{name}: {details['price']}")
```

Q no : 2 . 
Two sections submitted assignments:
section_a = {"Ali", "Sara", "Umar", "Hina", "Zara", "Kamran"}
section_b = {"Sara", "Bilal", "Hina", "Imran", "Zara", "Nadia"}

Using set operations (no loops), find:
•	Students in BOTH sections (intersection)
•	Students ONLY in Section A
•	All unique students across both sections (union)
•	Students in exactly ONE section (symmetric difference)

Also: create a tuple of the three highest-scoring subjects. Then demonstrate tuples are immutable by trying to change a value and catching the error with try/except.

```python
# --- SET OPERATIONS ---
section_a = {"Ali", "Sara", "Umar", "Hina", "Zara", "Kamran"}
section_b = {"Sara", "Bilal", "Hina", "Imran", "Zara", "Nadia"}

# 1. Students in BOTH sections (Intersection)
both_sections = section_a.intersection(section_b)
print(f"1. Both sections: {both_sections}")

# 2. Students ONLY in Section A (Difference)
only_section_a = section_a.difference(section_b)
print(f"2. Only in Section A: {only_section_a}")

# 3. All unique students across both (Union)
all_students = section_a.union(section_b)
print(f"3. Total unique students: {all_students}")

# 4. Students in exactly ONE section (Symmetric Difference)
unique_to_one = section_a.symmetric_difference(section_b)
print(f"4. Students in only one section: {unique_to_one}")

print("\n" + "-"*30 + "\n")

# --- TUPLE OPERATIONS ---

# 1. Create a tuple of three highest-scoring subjects
top_subjects = ("Python", "Math", "Physics")
print(f"Top Subjects: {top_subjects}")

# 2. Demonstrate Tuples are Immutable
print("\nAttempting to change a value in the tuple...")

try:
    # Trying to change "Math" to "English"
    top_subjects[1] = "English"
except TypeError as e:
    print(f"Error caught! You cannot change a tuple: {e}")
```

Part 1 — Book class (3 marks):
Create a Book class with:
•	Attributes: title, author, isbn, is_available (default True)
•	_str_ method that returns a readable string
•	checkout() and return_book() methods that toggle availability

Part 2 — Library class (4 marks):
Create a Library class that:
•	Stores a list of Book objects
•	add_book(book) — adds a book
•	find_by_author(author) — returns all books by an author
•	available_books() — returns all currently available books
•	total_books() — returns count of all books

Create at least 4 Book objects, add to Library, checkout 2, then display all available books.

```python
# Part 1: The Book Class
class Book:
    def __init__(self, title, author, isbn):
        self.title = title
        self.author = author
        self.isbn = isbn
        self.is_available = True  # Default value

    def checkout(self):
        """Toggles availability to False when a book is taken."""
        if self.is_available:
            self.is_available = False
            return True
        return False

    def return_book(self):
        """Toggles availability back to True."""
        self.is_available = True

    def __str__(self):
        status = "Available" if self.is_available else "Checked Out"
        return f"'{self.title}' by {self.author} (ISBN: {self.isbn}) - [{status}]"


# Part 2: The Library Class
class Library:
    def __init__(self):
        self.books = []

    def add_book(self, book):
        """Adds a Book object to the library list."""
        self.books.append(book)

    def find_by_author(self, author):
        """Returns a list of books by a specific author."""
        return [book for book in self.books if book.author.lower() == author.lower()]

    def available_books(self):
        """Returns a list of books that are currently available."""
        return [book for book in self.books if book.is_available]

    def total_books(self):
        """Returns the total count of books in the library."""
        return len(self.books)


# --- Execution and Testing ---

# 1. Create 4 Book objects
b1 = Book("The Great Gatsby", "F. Scott Fitzgerald", "9780743273565")
b2 = Book("Python Crash Course", "Eric Matthes", "9781593279288")
b3 = Book("To Kill a Mockingbird", "Harper Lee", "9780061120084")
b4 = Book("Automate the Boring Stuff", "Al Sweigart", "9781593275990")

# 2. Add books to the Library
my_library = Library()
my_library.add_book(b1)
my_library.add_book(b2)
my_library.add_book(b3)
my_library.add_book(b4)

# 3. Checkout 2 books
b1.checkout()
b2.checkout()

# 4. Display all available books
print(f"Total books in library: {my_library.total_books()}")
print("\n--- Available Books ---")
for book in my_library.available_books():
    print(book)
```

Write a script that manages a text-based student record file called students.txt.
•	save_student(name, marks) — appends a student record in format: Ali,85
•	load_students() — reads file and returns list of dicts: [{"name": "Ali", "marks": 85}, ...]
•	top_student() — reads file and returns name of student with highest marks

Save at least 5 students, then call top_student() and print the result.

```python
import os

def save_student(name, marks):
    """Appends a student record in format: Name,Marks"""
    with open("students.txt", "a") as file:
        file.write(f"{name},{marks}\n")

def load_students():
    """Reads the file and returns a list of dictionaries."""
    students_list = []
    if not os.path.exists("students.txt"):
        return students_list

    with open("students.txt", "r") as file:
        for line in file:
            # Strip removes the newline, split(',') separates name and marks
            parts = line.strip().split(",")
            if len(parts) == 2:
                students_list.append({
                    "name": parts[0],
                    "marks": int(parts[1])
                })
    return students_list

def top_student():
    """Reads the file and returns the name of the student with highest marks."""
    all_students = load_students()
    if not all_students:
        return "No records found."
    
    # Use max() with a key to find the highest marks
    best_student = max(all_students, key=lambda s: s["marks"])
    return best_student["name"]

# --- Execution ---

# 1. Save 5 students
records = [
    ("Ali", 85),
    ("Sara", 92),
    ("Umar", 78),
    ("Hina", 95),
    ("Bilal", 88)
]

# Clear existing file for a fresh start (optional)
if os.path.exists("students.txt"):
    os.remove("students.txt")

for name, marks in records:
    save_student(name, marks)

# 2. Find and print the top student
winner = top_student()
print(f"The top student is: {winner}")
```
