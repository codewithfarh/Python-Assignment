Q no : 1

Write the following standalone utility functions, each with a docstring:
•	is_prime(n) — returns True if n is prime, False otherwise
•	celsius_to_all(c) — returns a tuple of (fahrenheit, kelvin)
•	count_vowels(text) — returns the count of vowels in a string
•	power(base, exp=2) — returns base raised to exp. Default exponent is 2.

Call each function with at least one test case and print the result.

````python
def is_prime(n):
    """Returns True if n is prime, False otherwise."""
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

def celsius_to_all(c):
    """Returns a tuple of (fahrenheit, kelvin)."""
    fahrenheit = (c * 9/5) + 32
    kelvin = c + 273.15
    return (fahrenheit, kelvin)

def count_vowels(text):
    """Returns the count of vowels (a, e, i, o, u) in a string."""
    vowels = "aeiouAEIOU"
    count = 0
    for char in text:
        if char in vowels:
            count += 1
    return count

def power(base, exp=2):
    """Returns base raised to exp. Default exponent is 2."""
    return base ** exp

# --- Test Cases ---

# Testing is_prime
print(f"Is 17 prime ? {is_prime(17)}")
print(f"Is 4 prime? {is_prime(4)}")

# Testing celsius_to_all
f, k = celsius_to_all(25)
print(f"25°C is {f}°F and {k}K")

# Testing count_vowels
sample_text = "Python Programming"
print(f"Vowels in '{sample_text}': {count_vowels(sample_text)}")

# Testing power
print(f"5 squared (default): {power(5)}")
print(f"2 to the power of 3: {power(2, 3)}")
````

````python
def total_marks(*args):
    """Accepts any number of integers and returns their sum."""
    return sum(args)

def generate_report(**student_info):
    """Accepts student data as keyword arguments and prints a formatted report card."""

    name = student_info.get("name", "N/A")
    roll_no = student_info.get("roll_no", "N/A")
    subjects = student_info.get("subjects", [])
    marks = student_info.get("marks", [])

    if marks:
        avg = sum(marks) / len(marks)
    else:
        avg = 0


    if avg >= 90:
        grade = "A+ (Outstanding)"
    elif avg >= 80:
        grade = "A (Excellent)"
    elif avg >= 70:
        grade = "B (Good)"
    elif avg >= 60:
        grade = "C (Average)"
    elif avg >= 50:
        grade = "D (Pass)"
    else:
        grade = "F (Fail)"

    print("-" * 30)
    print("      STUDENT REPORT CARD")
    print("-" * 30)
    print(f"Name:      {name}")
    print(f"Roll No:   {roll_no}")
    print(f"Subjects:  {', '.join(subjects)}")
    print(f"Average:   {avg:.2f}")
    print(f"Grade:     {grade}")
    print("-" * 30)

# --- Test Calls ---


total = total_marks(85, 78, 92, 10, 5)
print(f"Sum of marks: {total}")


generate_report(
    name="Sana Khan",
    roll_no="CS-101",
    subjects=["Math", "Physics", "Python"],
    marks=[85, 78, 92]
)
````

    {"name": "Umar",  "marks": 45, "city": "Karachi"},

    {"name": "Bilal", "marks": 63, "city": "Lahore"},
]

Using lambda, map, filter, and sorted — NO for loops:
•	Create a list of just names using map
•	Filter only students who passed (marks >= 50) using filter
•	Sort students by marks descending using sorted
•	Use map to create list of strings like "Ali - 55 marks"
•	Use filter to get only students from Karachi

Print the result of each operation.

````python
students = [
    {"name": "Ali",   "marks": 55, "city": "Karachi"},
    {"name": "Sara",  "marks": 90, "city": "Lahore"},
    {"name": "Umar",  "marks": 45, "city": "Karachi"},
    {"name": "Hina",  "marks": 78, "city": "Islamabad"},
    {"name": "Bilal", "marks": 63, "city": "Lahore"},
]

# 1. Create a list of just names using map
student_names = list(map(lambda s: s["name"], students))
print("1. Student Names:", student_names)

# 2. Filter only students who passed (marks >= 50)
passed_students = list(filter(lambda s: s["marks"] >= 50, students))
print("2. Passed Students:", passed_students)

# 3. Sort students by marks descending
sorted_students = sorted(students, key=lambda s: s["marks"], reverse=True)
print("3. Sorted by Marks (Descending):", sorted_students)

# 4. Use map to create list of strings like "Ali - 55 marks"
formatted_list = list(map(lambda s: f"{s['name']} - {s['marks']} marks", students))
print("4. Formatted Strings:", formatted_list)

# 5. Use filter to get only students from Karachi
karachi_students = list(filter(lambda s: s["city"] == "Karachi", students))
print("5. Karachi Students:", karachi_students)
````

Write fibonacci(n) recursively. Print the first 10 Fibonacci numbers.

Part 2 — Factorial (2 marks):

Part 3 — Flatten a nested list (2 marks):
Write flatten(lst) that takes a deeply nested list and returns a flat list using recursion.
flatten([1, [2, [3, 4], 5], [6, 7]])
# Expected: [1, 2, 3, 4, 5, 6, 7]

````python
def fibonacci(n):
    """Returns the nth Fibonacci number recursively."""
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)

# Print the first 10 Fibonacci numbers
print("First 10 Fibonacci numbers:")
for i in range(10):
    print(fibonacci(i), end=" ")
print()

def factorial(n):
    """Returns the factorial of n recursively."""
    if n < 0:
        return "Error: Negative numbers not allowed"
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)

# Test cases
print(f"Factorial of 5: {factorial(5)}")
print(f"Factorial of 0: {factorial(0)}")
print(f"Factorial of -5: {factorial(-5)}")


def flatten(lst):
    """Flattens a deeply nested list using recursion."""
    flat_list = []
    for item in lst:
        if isinstance(item, list):
            # Recursively flatten the sub-list and extend our result
            flat_list.extend(flatten(item))
        else:
            flat_list.append(item)
    return flat_list

# Test the function
nested = [1, [2, [3, 4], 5], [6, 7]]
print(f"Original: {nested}")
print(f"Flattened: {flatten(nested)}")
````
